# Simple Theory About Pytorch
A simple guide to underestand pytorch

```python
#imports
import torch
import numpy as np
```



| Dimensión | Nombre común | Ejemplo | Forma (shape) |
|-----------|--------------|---------|---------------|
| 0 | Escalar | `42` | `()` |
| 1 | Vector | `[1, 2, 3]` | `(3,)` |
| 2 | Matriz | `[[1,2],[3,4]]` | `(2,2)` |
| 3 | Tensor 3D | Imagen RGB | `(3, 480, 640)` |
| 4 | Tensor 4D | Batch de imágenes | `(32, 3, 480, 640)` |
| 5+ | Tensor ND | Secuencias de video | `(batch, frames, canales, alto, ancho)` |

contenedores de diferentes dimensiones:
- Escalar: una caja con un solo número
- Vector: un tubo con números en fila
- Matriz: una bandeja con filas y columnas
- Tensor 3D: un cubo de Rubik con números en cada celda
- Tensor 4D: varios cubos de Rubik apilados



```python
# create bidimensional tensor
tensor_bi = torch.zeros(3,5)
#tensor_bi = torch.ones(3,5)    # para crear un tensor con unos
print(tensor_bi)

# create a 2x3x4 tesor
tensor_tri = torch.rand(2,3,4)
print(tensor_tri)

# tensor desde una lista de datos
data = [[1,2,3], [4,5,6]]
converted = torch.tensor(data)
```
